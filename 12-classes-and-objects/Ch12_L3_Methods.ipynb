### 1. Defining a Method
A method is a function defined inside a class that operates on `self` (the instance).

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        return f"Hi, I'm {self.name} and I'm {self.age} years old."

me = Person("Sara", 21)
print(me.introduce())
```

### 2. A Method That Changes State

```python
class Counter:
    def __init__(self):
        self.count = 0

    def increment(self):
        self.count += 1

c = Counter()
c.increment()
print(c.count)
```
