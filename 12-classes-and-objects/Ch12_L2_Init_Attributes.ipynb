# `__init__` and Attributes

### 1. The Constructor Method
`__init__` runs automatically when you create a new instance, letting you set up attributes immediately — `self` refers to "this particular instance".

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

me = Person("Sara", 21)
print(me.name, me.age)
```

### 2. Default Parameter Values in `__init__`

```python
class Goal:
    def __init__(self, title, status="not started"):
        self.title = title
        self.status = status

g1 = Goal("Learn Python")
g2 = Goal("Run a 5k", status="in progress")
print(g1.title, "-", g1.status)
print(g2.title, "-", g2.status)
```
