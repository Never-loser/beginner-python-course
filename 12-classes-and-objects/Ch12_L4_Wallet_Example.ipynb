# A Practical Example: the `Wallet` Class

### Putting It All Together
This class combines `__init__`, attributes, methods that change state, default values, and (from Chapter 10) a custom validation check using `raise`.

```python
class Wallet:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError(f"Cannot withdraw {amount}: balance is only {self.balance}.")
        self.balance -= amount

    def show_balance(self):
        print(f"{self.owner}'s balance: {self.balance}")

w = Wallet("Sara", 100)
w.deposit(50)
w.show_balance()
w.withdraw(30)
w.show_balance()
```
