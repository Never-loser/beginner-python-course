# Classes and Objects

### 1. Class = Blueprint, Object = Instance
A **class** is a blueprint; an **object** (or **instance**) is a real thing built from that blueprint. You can create many different objects from the same class.

```python
class Dog:
    pass

my_dog = Dog()
your_dog = Dog()
print(type(my_dog))
print(my_dog is your_dog)   # two separate objects, even though same class
```

### 2. Objects Can Have Attributes Added Freely
Unlike many other languages, Python lets you attach a new attribute to a specific instance on the fly.

```python
my_dog = Dog()
my_dog.name = "Rex"
print(my_dog.name)
# print(your_dog.name)   # would crash — your_dog never got a "name"
```
