### 1. What is a Set?
A **set** is an **unordered** collection of **unique** values, written with curly braces `{ }`. Duplicates are automatically removed.

```python
numbers = {1, 2, 2, 3, 3, 3}
print(numbers)     # duplicates gone: {1, 2, 3}
print(type(numbers))
```

### 2. Creating an Empty Set
`{}` alone creates an empty **dictionary**, not a set (we'll see dictionaries next lesson). For an empty set, you must use `set()`.

```python
empty_set = set()
print(type(empty_set))
```

### 3. Sets Are Unordered — No Indexing
Because a set has no fixed order, you cannot access an element by position like `my_set[0]`.

```python
# This will crash on purpose — read the error message
colors = {"red", "green", "blue"}
print(colors[0])
```

### 4. Set Operations

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(a | b)   # union — everything in either
print(a & b)   # intersection — only what's in both
print(a - b)   # difference — in a, but not in b
```
