# Dictionaries — Definition and Access

### 1. What is a Dictionary?
A **dictionary** stores **key–value pairs**, written `{key: value, ...}`. Instead of accessing by position (like a list), you access by **key**.

````python
student = {"name": "Sara", "age": 21, "city": "Tehran"}
print(student["name"])
print(student["age"])
````

### 2. Accessing a Missing Key
Accessing a key that doesn't exist raises a `KeyError`. Use `.get(key, default)` to safely access a key that might not exist, without crashing.

````python
student = {"name": "Sara", "age": 21}
# print(student["email"])          # would crash — uncomment to try
print(student.get("email"))        # prints None instead of crashing
print(student.get("email", "not provided"))   # or a custom default
````

### 3. Adding and Updating Keys

````python
student = {"name": "Sara", "age": 21}
student["age"] = 22          # update existing key
student["city"] = "Tehran"   # add a new key
print(student)
````

### 4. Checking Membership With `in`
For a dictionary, `in` checks the **keys**, not the values!

````python
student = {"name": "Sara", "age": 21}
print("name" in student)     # True — "name" is a key
print("Sara" in student)     # False — "Sara" is a VALUE, not a key
````
