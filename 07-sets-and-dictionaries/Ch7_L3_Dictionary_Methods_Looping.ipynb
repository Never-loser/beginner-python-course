# Dictionary Methods and Looping

### 1. `.keys()`, `.values()`, `.items()`

````python
student = {"name": "Sara", "age": 21, "city": "Tehran"}
print(student.keys())
print(student.values())
print(student.items())
````

### 2. Looping Over a Dictionary
Looping directly over a dictionary gives you the **keys**. Use `.items()` to get both key and value together.

````python
student = {"name": "Sara", "age": 21, "city": "Tehran"}

for key in student:
    print(key)

print("---")

for key, value in student.items():
    print(key, "->", value)
````

### 3. `.update()` and Removing Keys

````python
a = {"name": "Sara", "age": 21}
b = {"age": 22, "city": "Tehran"}   # note: "age" appears in both
a.update(b)
print(a)   # what happens to "age"?

a.pop("city")
print(a)
````
