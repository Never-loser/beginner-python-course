# f-strings and Formatting

### 1. Why f-strings?
Until now we built output using commas in `print()`. **f-strings** (formatted string literals) let you embed variables and expressions **directly inside** a string, which is cleaner and more powerful. Put an `f` right before the opening quote, and wrap variables/expressions in `{ }`.

````python
name = "Sara"
age = 21
print(f"Hi, I'm {name} and I am {age} years old.")
print(f"Next year I will be {age + 1}.")   # expressions work too!
````

### 2. Formatting Numbers Inside f-strings
Add a `:` followed by a format spec after the variable name:
- `:.2f` → 2 decimal places
- `:,` → thousands separator
- `:>10` / `:<10` → right/left-align within 10 characters (useful for tables)

````python
price = 1234.5678
print(f"{price:.2f}")     # 1234.57
print(f"{price:,.2f}")    # 1,234.57
print(f"[{'hi':>10}]")     # right-aligned in 10 spaces
print(f"[{'hi':<10}]")     # left-aligned in 10 spaces
````
