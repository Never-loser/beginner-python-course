# Common String Methods

### 1. Case Methods

````python
s = "  Hello Python  "
print(s.upper())
print(s.lower())
print(s.title())
````

### 2. Cleaning Whitespace

````python
s = "  Hello Python  "
print("[" + s.strip() + "]")    # removes leading/trailing spaces only
````

### 3. Searching and Replacing

````python
s = "I love pizza. Pizza is great."
print(s.replace("Pizza", "Sushi"))
print(s.find("pizza"))    # index of first match, -1 if not found
print(s.count("izza"))    # how many times it appears
````

### 4. Splitting and Joining

````python
sentence = "python is fun"
words = sentence.split()      # splits by whitespace into a list
print(words)
print("-".join(words))        # joins a list back into one string
````

### 5. Checking Start/End

````python
filename = "report.pdf"
print(filename.startswith("report"))
print(filename.endswith(".pdf"))
````
