# Strings and Indexing

### 1. What is a String?
A **string** (`str`) is a sequence of characters, written inside single `'...'`, double `"..."`, or triple `'''...'''`/`"""..."""` quotes (triple quotes allow multi-line text).

````python
greeting = "Hello, World!"
print(greeting)
print(len(greeting))   # number of characters, including spaces and punctuation
````

### 2. Indexing
Every character in a string has a position (**index**), starting at `0` from the left. You can also count from the **end** using negative numbers, starting at `-1`.

````python
word = "Python"
print(word[0])    # first character
print(word[5])    # last character
print(word[-1])   # also the last character
print(word[-2])   # second-to-last
````

### 3. Strings Are Immutable
Once created, you **cannot** change a single character of a string in place — trying to assign to an index raises an error. To "change" a string, you create a new one.

````python
# This will crash on purpose — read the error message
word = "Python"
word[0] = "J"
````
