# Slicing Strings

### 1. Slice Syntax: `string[start:stop:step]`
- `start` — index to begin at (inclusive). Default: beginning.
- `stop` — index to stop **before** (exclusive). Default: end.
- `step` — how many characters to skip each time. Default: `1`.

````python
word = "Python Programming"
print(word[0:6])     # 'Python'
print(word[7:])      # everything from index 7 to the end
print(word[:6])      # everything up to (not including) index 6
````

### 2. Negative Slicing and Step
Negative indices and a negative step both work — a step of `-1` reverses the string.

````python
word = "Python"
print(word[-6:-1])
print(word[::-1])     # reverses the whole string
print(word[::2])      # every 2nd character
````

### 3. A Useful Gotcha: Slices Don't Crash on Out-of-Range Indices
Unlike direct indexing (`word[99]`, which raises an error), slicing with an out-of-range index simply returns as much as exists — no error.

````python
word = "Python"
print(word[2:100])   # no error, even though the string is shorter than 100
````
