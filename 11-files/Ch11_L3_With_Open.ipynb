# Using `with open()` — the Right Way

### 1. Why `with`?
`with open(...) as f:` automatically closes the file for you — even if an error happens inside the block. This is the standard, recommended way to work with files in real Python code.

```python
with open("my_info.txt", "r") as f:
    content = f.read()
print(content)
# the file is already closed here — no need to call f.close()
```
