### 2. Opening and Reading a File

```python
f = open("my_info.txt", "r")   # "r" = read mode
content = f.read()
print(content)
f.close()                      # always close when done (Lesson 3 shows a better way)
```

### 3. `.readlines()` vs `.read()`

```python
f = open("my_info.txt", "r")
lines = f.readlines()   # returns a LIST of lines (each ending in \n)
print(lines)
f.close()
```
