# Writing and Appending to Files

### 1. `"w"` Mode — Write (Overwrites!)
Opening a file in `"w"` mode creates it if it doesn't exist, or **completely erases** its previous content if it does.

```python
f = open("diary.txt", "w")
f.write("Today I started learning about files in Python.\n")
f.close()

f = open("diary.txt", "r")
print(f.read())
f.close()
```

### 2. `"a"` Mode — Append (Adds to the End)

```python
f = open("diary.txt", "a")
f.write("This line was added without erasing the first one.\n")
f.close()

f = open("diary.txt", "r")
print(f.read())
f.close()
```
