## 🇬🇧 English

### 1. What is a Notebook?
A **notebook** (`.ipynb` file) is made of **cells** you run one at a time, mixing code, formatted text, and output — all in one document. This is different from a plain `.py` script, which runs top to bottom in one go.

### 2. Two Types of Cells
- **Code cells** — run Python code.
- **Markdown cells** — formatted text, like this one (headers, **bold**, *italic*, lists...).

You can change a cell's type from the toolbar or the dropdown near the top of each cell.

### 3. Running Cells
- `Shift + Enter` → run the current cell, move to the next one.
- `Ctrl + Enter` → run the current cell, stay on it.
- `Esc` then `B` → insert a new cell below. `Esc` then `A` → insert above.

Try it in the empty code cell below this text.

````python
# Try running this cell with Shift+Enter
print("Hello, Jupyter!")
````

### 4. Execution Order — the `In [n]:` Number
Jupyter runs cells in the **order you run them**, not automatically top-to-bottom! The number in `In [n]:` next to a code cell shows *when* (in what order) you ran it — not its position in the file.

This means if you run cell 3, then cell 1, then cell 2, variables will behave according to that real order — which can create confusing bugs if you're not careful.

### 5. The Kernel
The **kernel** is the running "engine" behind your notebook. It holds all your variables in memory as you run cells. If you:
- **Restart the kernel** → all variables are wiped; you start fresh.
- **Interrupt the kernel** → stop a cell that's stuck (e.g. an infinite loop).

You'll find these options under the "Restart" / "Interrupt" buttons at the top of the notebook interface.

### 7. Saving
`Ctrl+S` (`Cmd+S` on Mac) saves the `.ipynb` file, including all outputs.

### ۱. نوت‌بوک چیست؟
یک **نوت‌بوک** (فایل `.ipynb`) از **سلول‌هایی** تشکیل شده که یکی‌یکی اجرا می‌کنید و می‌توانید در آن کد، متن فرمت‌شده و خروجی را با هم ترکیب کنید — همه در یک فایل. این با یک اسکریپت ساده‌ی `.py` فرق دارد که یک‌جا از بالا تا پایین اجرا می‌شود.

### ۲. دو نوع سلول
- **سلول کد (Code)** — کد پایتون را اجرا می‌کند.
- **سلول مارک‌داون (Markdown)** — متن فرمت‌شده، مثل همین سلول (تیتر، **بولد**، *ایتالیک*، لیست...).

می‌توانید نوع سلول را از نوار ابزار یا منوی کشویی بالای هر سلول تغییر دهید.

### ۳. اجرای سلول‌ها
- `Shift + Enter` → اجرای سلول فعلی، رفتن به سلول بعدی.
- `Ctrl + Enter` → اجرای سلول فعلی، ماندن روی همان سلول.
- `Esc` سپس `B` → افزودن سلول جدید در پایین. `Esc` سپس `A` → افزودن در بالا.

در سلول کد خالیِ زیر همین متن امتحان کنید.

````python
# این سلول را با Shift+Enter اجرا کنید
print("سلام، ژوپیتر!")
````

### ۴. ترتیب اجرا — عدد `In [n]:`
ژوپیتر سلول‌ها را به **همان ترتیبی که شما اجرا می‌کنید** اجرا می‌کند، نه لزوماً از بالا به پایین! عددِ داخل `In [n]:` کنار هر سلول کد نشان می‌دهد *چه‌زمانی* (به چه ترتیبی) آن را اجرا کرده‌اید — نه جایگاهش در فایل.

یعنی اگر سلول ۳ را اجرا کنید، بعد سلول ۱، بعد سلول ۲، متغیرها طبق همین ترتیب واقعی رفتار می‌کنند — که اگر دقت نکنید می‌تواند باگ‌های گیج‌کننده‌ای ایجاد کند.

### ۵. کرنل (Kernel)
**کرنل** موتور در حال اجرا پشت نوت‌بوک شماست. تمام متغیرهایتان را در حافظه نگه می‌دارد وقتی سلول‌ها را اجرا می‌کنید. اگر:
- **کرنل را ری‌استارت کنید** → همه‌ی متغیرها پاک می‌شوند و از نو شروع می‌کنید.
- **کرنل را متوقف (Interrupt) کنید** → سلولی که گیر کرده (مثلاً حلقه‌ی بی‌نهایت) متوقف می‌شود.

این گزینه‌ها را زیر دکمه‌های «Restart» / «Interrupt» در بالای محیط نوت‌بوک پیدا می‌کنید.

### تمرین‌ها

**🟢 آسان**
در سلول کد زیر، با `print()` برنامه‌ای بنویسید که یک خوش‌آمدگویی کوتاه شامل **نام و نام‌خانوادگی واقعی خودتان** چاپ کند (مثلاً: `سلام، من سارا احمدی هستم و دارم پایتون یاد می‌گیرم!`). سپس در یک سلول مارک‌داون بالای آن، یک بیوی کوتاه بنویسید: نامتان، یک سرگرمی، و اینکه چرا پایتون یاد می‌گیرید.
> چون باید شامل *نام واقعی خودتان* باشد، این تمرین بین هم‌کلاسی‌ها قابل کپی نیست.

**🟡 متوسط**
در ادامه، **۴ سلول کد کوتاه** بسازید. هرکدام یک متغیر را تعریف یا تغییر دهد (مثلاً سلول A مقدار `x = 1` را ست کند، سلول B مقدار `x = x + 10` را). آن‌ها را **عمداً خارج از ترتیب** اجرا کنید (نه از بالا به پایین). از عددهای `In [ ]:` اسکرین‌شات بگیرید که ترتیب واقعی اجرای شما را نشان دهد. در یک سلول مارک‌داون، در ۲ تا ۳ جمله توضیح دهید چرا اگر سلول‌ها را دقیقاً از بالا به پایین اجرا می‌کردید، مقدار نهایی متغیر فرق می‌کرد.
> عددهای `In [n]:` فقط از *اجرای واقعی* سلول‌ها در ژوپیتر ظاهر می‌شوند — یک پاسخ صرفاً متنی نمی‌تواند این را جعل کند.
