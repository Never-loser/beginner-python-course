## 🇬🇧 English

### 1. The Simplest Program
`print()` displays (outputs) something on the screen. It is usually the very first thing anyone learns in any programming language.

````python
print("Hello, World!")
````

### 2. Printing Text vs Numbers
Text (a "string") goes inside quotes `"..."` or `'...'`. Numbers do not need quotes.

````python
print("My age is:")
print(25)
````

### 3. Printing Multiple Values at Once
Separate values with commas — `print()` will put a space between them automatically.

````python
print("Age:", 25, "City:", "Tehran")
````

### 4. The `sep` Parameter
`sep` controls what is placed *between* values (default is a single space).

````python
print("2026", "08", "27", sep="-")   # prints 2026-08-27
````

### 5. The `end` Parameter
`end` controls what is placed *after* the whole print statement (default is a newline `\n`, which is why each `print()` normally starts on a new line).

````python
print("Loading", end="...")
print("Done!")
````

### 6. Escape Characters
Special characters inside strings, written with a backslash:
- `\n` → new line
- `\t` → tab (horizontal space)
- `\"` → a literal quotation mark inside a double-quoted string
- `\\` → a literal backslash

We'll cover proper string formatting (f-strings) in Chapter 4 — for now, plain `print()` with commas is enough.

> Note: this lesson intentionally keeps things minimal. We are *not* using f-strings or variables yet — that comes in later chapters.

````python
print("Line one\nLine two")
print("Name:\tAli")
````

## 🇮🇷 فارسی

### ۱. ساده‌ترین برنامه
`print()` چیزی را روی صفحه نمایش می‌دهد (چاپ می‌کند). معمولاً اولین چیزی است که در هر زبان برنامه‌نویسی یاد گرفته می‌شود.

````python
print("سلام دنیا!")
````

### ۲. چاپ متن در مقابل چاپ عدد
متن (یک «رشته» / string) داخل گیومه `"..."` یا `'...'` قرار می‌گیرد. عددها نیازی به گیومه ندارند.

````python
print("سن من:")
print(25)
````

### ۳. چاپ چند مقدار با هم
مقادیر را با کاما جدا کنید — `print()` به‌طور خودکار بین آن‌ها یک فاصله می‌گذارد.

````python
print("سن:", 25, "شهر:", "تهران")
````

### ۴. پارامتر `sep`
`sep` مشخص می‌کند *بین* مقادیر چه چیزی قرار بگیرد (پیش‌فرض یک فاصله است).

````python
print("2026", "08", "27", sep="-")   # چاپ می‌کند: 2026-08-27
````

### ۵. پارامتر `end`
`end` مشخص می‌کند *بعد از* کل خروجی چاپ چه چیزی قرار بگیرد (پیش‌فرض یک خط جدید `\n` است، به همین دلیل هر `print()` معمولاً در یک خط جدید شروع می‌شود).

````python
print("در حال بارگذاری", end="...")
print("تمام شد!")
````

### ۶. کاراکترهای فرار (Escape Characters)
کاراکترهای خاص داخل رشته‌ها، با یک بک‌اسلش نوشته می‌شوند:
- `\n` → خط جدید
- `\t` → تب (فاصله‌ی افقی)
- `\"` → یک علامت گیومه‌ی واقعی داخل رشته‌ای که با گیومه‌ی دوتایی نوشته شده
- `\\` → یک بک‌اسلش واقعی

فرمت‌دهی درست رشته‌ها (f-string) را در فصل ۴ یاد می‌گیریم — فعلاً همین `print()` ساده به‌همراه کاما کافی است.

> نکته: این درس عمداً حداقلی نگه داشته شده. هنوز از f-string یا متغیر استفاده نمی‌کنیم — این‌ها در فصل‌های بعدی می‌آیند.

````python
print("خط اول\nخط دوم")
print("نام:\tعلی")
````
