# `elif` and Multiple Conditions

### 1. Chaining Conditions with `elif`
`elif` ("else if") lets you check several conditions in order. Python checks them **top to bottom** and runs the **first** one that's `True` — then skips the rest, even if a later condition would also be `True`.

```python
score = 85
if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
else:
    print("F")
```

### 2. Order Matters!
Because Python stops at the first `True` branch, putting conditions in the **wrong order** can create a logic bug that doesn't crash — it just silently gives the wrong answer.

```python
score = 95

# BUGGY order — the first condition catches everything above 70!
if score >= 70:
    print("C")
elif score >= 80:
    print("B")
elif score >= 90:
    print("A")
```
