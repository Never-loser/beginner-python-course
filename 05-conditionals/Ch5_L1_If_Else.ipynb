# `if` and `else`

### 1. Basic `if`
An `if` statement runs a block of code **only if** its condition is `True`. The indented block (4 spaces, by convention) is what belongs to the `if`.

```python
age = 20
if age >= 18:
    print("You are an adult.")
```

### 2. Adding `else`
`else` runs when the condition is `False` — it has no condition of its own, and always comes after an `if` (or `elif`) block.

```python
age = 15
if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")
```

### 3. Indentation Is Not Optional
Python uses indentation (not `{ }` or `end`) to know what belongs inside the `if` block. Inconsistent indentation causes an error.

```python
# This will crash on purpose — read the error message
age = 20
if age >= 18:
print("You are an adult.")
```
