# Combining Conditions with Logical Operators

### 1. Using `and` / `or` Inside `if`
You can combine multiple conditions from Chapter 3 directly inside an `if`.

```python
age = 20
has_ticket = True
if age >= 18 and has_ticket:
    print("You may enter.")
else:
    print("You may not enter.")
```

### 2. Nested `if` — Another Way to Combine Conditions
An `if` inside another `if` is called **nesting**. A nested pair of `if`s with `and` between them behaves the same as one combined condition — but reads differently.

```python
age = 20
has_ticket = True

# Same logic as above, written with nesting instead of "and"
if age >= 18:
    if has_ticket:
        print("You may enter.")
    else:
        print("You may not enter.")
else:
    print("You may not enter.")
```
