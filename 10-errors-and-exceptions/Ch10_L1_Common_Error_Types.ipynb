# Common Error Types in Python

### A Quick Reference
You've already met most of these in earlier chapters:
| Error | When it happens |
|---|---|
| `SyntaxError` | invalid Python grammar (Ch1 L2, Ch5 L1) |
| `NameError` | using a variable that doesn't exist |
| `TypeError` | wrong type for an operation (Ch1 L4, Ch4 L1) |
| `ValueError` | right type, but an invalid value (Ch2 L3) |
| `IndexError` | list/string index out of range (Ch4 L1) |
| `KeyError` | dictionary key doesn't exist (Ch7 L2) |
| `ZeroDivisionError` | dividing by zero |
