# Parameters, Arguments, and Default Values

### 1. Positional and Keyword Arguments

```python
def introduce(name, age):
    return f"{name} is {age} years old."

print(introduce("Sara", 21))          # positional
print(introduce(age=21, name="Sara")) # keyword — order doesn't matter here
```

### 2. Default Parameter Values
A default is used when the caller doesn't provide that argument.

```python
def introduce(name, age=18):
    return f"{name} is {age} years old."

print(introduce("Sara", 21))
print(introduce("Ali"))   # uses the default age
```
