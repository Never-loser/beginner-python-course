# Lambda Functions

### 1. What is a Lambda?
A **lambda** is a small, unnamed, one-expression function — useful for short throwaway logic, especially as a `key` for sorting.

```python
double = lambda x: x * 2
print(double(5))

# Equivalent regular function, for comparison:
def double_v2(x):
    return x * 2
```

### 2. Lambdas With `sorted()`

```python
people = [("Sara", 21), ("Ali", 30), ("Reza", 18)]
people.sort(key=lambda person: person[1])   # sort by age (2nd item)
print(people)
```
