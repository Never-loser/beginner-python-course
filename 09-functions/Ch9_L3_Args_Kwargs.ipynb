# `*args` and `**kwargs`

### 1. `*args` — Any Number of Positional Arguments
Collects extra positional arguments into a **tuple**.

```python
def total(*args):
    print(args)        # see what it actually is
    return sum(args)

print(total(1, 2, 3))
print(total(10, 20))
```

### 2. `**kwargs` — Any Number of Keyword Arguments
Collects extra keyword arguments into a **dictionary**.

```python
def profile(**kwargs):
    print(kwargs)

profile(name="Sara", age=21, city="Tehran")
```
