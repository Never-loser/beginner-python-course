# Defining Functions and `return`

### 1. Why Functions?
A function packages a block of code under a name so you can reuse it without retyping it.

```python
def greet():
    print("Hello!")

greet()
greet()   # call it as many times as you want
```

### 2. `return` vs `print()` — a Critical Difference
`print()` only **displays** something; the function itself gives back `None` to any code that called it. `return` actually **hands the value back** so it can be stored, reused, or combined with other code.

```python
def add_v1(a, b):
    print(a + b)     # only displays

def add_v2(a, b):
    return a + b      # actually gives the value back

result1 = add_v1(2, 3)
result2 = add_v2(2, 3)
print("result1 is:", result1)   # None!
print("result2 is:", result2)   # 5
```
