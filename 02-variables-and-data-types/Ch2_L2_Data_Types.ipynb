# Basic Data Types

### 1. The Four Basic Types
| Type | Meaning | Example |
|---|---|---|
| `int` | whole number | `42` |
| `float` | decimal number | `3.14` |
| `str` | text | `"hello"` |
| `bool` | True / False | `True` |

### 2. Checking a Value's Type
Use the built-in `type()` function.

````python
print(type(42))
print(type(3.14))
print(type("hello"))
print(type(True))
````

### 3. Python Figures Out the Type Automatically
You never declare a type — Python infers it from the value you assign. This is called **dynamic typing**.

### 4. Booleans Come From Comparisons Too

````python
print(5 > 3)
print(5 == 3)
print(type(5 > 3))
````
