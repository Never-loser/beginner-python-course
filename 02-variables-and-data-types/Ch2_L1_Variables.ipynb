# Variables and Naming Rules

### 1. What is a Variable?
A **variable** is a name that refers to a value stored in memory. In Python, you create one simply by assigning a value with `=` — no need to declare a type first.

````python
name = "Sara"
age = 21
print(name)
print(age)
````

### 2. Naming Rules
- Must start with a letter or underscore `_` — **not** a digit.
- Can contain letters, digits, and underscores after that.
- Case-sensitive: `age` and `Age` are different variables.
- Cannot be a Python **reserved keyword** (`class`, `for`, `if`, `True`, ...).
- Convention: use `snake_case` (lowercase words separated by underscores), e.g. `student_name`.

### 3. Reassignment
A variable can be reassigned to a new value (even a different type) at any time — Python is **dynamically typed**.

````python
x = 5
print(x)
x = "now I am a string"
print(x)
````

### 4. Multiple Assignment

````python
a, b, c = 1, 2, 3
print(a, b, c)
````
