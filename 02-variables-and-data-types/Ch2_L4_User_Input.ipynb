# Getting Input From the User

### 1. The `input()` Function
`input()` pauses your program, waits for the user to type something and press Enter, and returns **whatever they typed — always as a string.**

```python
name = input("What is your name? ")
print("Hello,", name)
```

### 2. Input Is Always a String — Even Numbers
If you ask for a number with `input()`, you must cast it yourself before doing math with it.

```python
# This looks like it should work, but it does NOT add numbers — try it!
age_text = input("How old are you? ")
print(age_text + 1)   # this line will crash — read the error
```

```python
# The fixed version
age_text = input("How old are you? ")
age = int(age_text)
print(age + 1)
```
