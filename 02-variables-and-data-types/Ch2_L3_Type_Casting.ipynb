# Type Casting (Converting Between Types)

### 1. Why We Need Casting
Sometimes a value is the "wrong" type for what you want to do with it — e.g. you have a number as text and need it as an actual number. Python lets you **cast** (convert) between types using `int()`, `float()`, `str()`, and `bool()`.

```python
age = 21
age_text = str(age)
print(age_text, type(age_text))

height_text = "1.75"
height_number = float(height_text)
print(height_number, type(height_number))
```

### 2. Casting Can Fail
Not every conversion makes sense. Casting a non-numeric string to a number raises an error.

```python
# This line will crash on purpose — read the error message
int("hello")
```

### 3. Truthiness — `bool()` on Other Types
Every value in Python is "truthy" or "falsy" when converted to `bool`. Generally: `0`, `0.0`, `""` (empty string), and empty containers are `False`; almost everything else is `True`.
