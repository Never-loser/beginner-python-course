# Logical Operators and Assignment Operators

### 1. Logical Operators
| Operator | Meaning |
|---|---|
| `and` | `True` only if **both** sides are `True` |
| `or` | `True` if **at least one** side is `True` |
| `not` | flips `True`/`False` |

These are usually combined with comparison operators from Lesson 2.

````python
age = 20
has_id = True
print(age >= 18 and has_id)
print(age >= 18 or has_id)
print(not has_id)
````

### 2. Precedence: `not` > `and` > `or`
Just like arithmetic has an order, so does logic. Use parentheses when in doubt.

````python
print(True or False and False)     # and runs first
print((True or False) and False)   # parentheses change the result
````

### 3. Augmented Assignment Operators
A shortcut for "update a variable based on its own current value":
| Shortcut | Same as |
|---|---|
| `x += 1` | `x = x + 1` |
| `x -= 1` | `x = x - 1` |
| `x *= 2` | `x = x * 2` |
| `x /= 2` | `x = x / 2` |
| `x //= 2` | `x = x // 2` |
| `x %= 2` | `x = x % 2` |
| `x **= 2` | `x = x ** 2` |

````python
score = 10
score += 5
print(score)
score *= 2
print(score)
````
