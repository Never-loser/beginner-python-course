# Comparison Operators

### 1. The Comparison Operators
| Operator | Meaning |
|---|---|
| `==` | equal to |
| `!=` | not equal to |
| `>` | greater than |
| `<` | less than |
| `>=` | greater than or equal to |
| `<=` | less than or equal to |

Every comparison produces a `bool` (`True` or `False`).

````python
print(5 == 5)
print(5 != 3)
print(5 > 10)
````

### 2. Comparing Different Types
Comparing a number to a string with `==` does **not** raise an error — it's simply always `False`, because they're different types. Strings are compared **alphabetically** (technically, by character codes), not by length.

````python
print(10 == "10")     # False — different types
print("Apple" < "apple")   # capital letters come before lowercase letters
print("10" < "9")     # comparing as TEXT, not as numbers!
````

### 3. Python Lets You Chain Comparisons

````python
x = 7
print(1 < x < 10)   # same as: 1 < x and x < 10
````
