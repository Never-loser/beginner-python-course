# Arithmetic Operators

### 1. The Basic Operators
| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` | addition | `5 + 2` | `7` |
| `-` | subtraction | `5 - 2` | `3` |
| `*` | multiplication | `5 * 2` | `10` |
| `/` | division (always returns float) | `5 / 2` | `2.5` |
| `//` | floor division (rounds down) | `5 // 2` | `2` |
| `%` | modulo (remainder) | `5 % 2` | `1` |
| `**` | exponent (power) | `5 ** 2` | `25` |

````python
print(5 + 2)
print(5 / 2)
print(5 // 2)
print(5 % 2)
print(5 ** 2)
````

### 2. Order of Operations (Precedence)
Python follows standard math precedence: `**` first, then `*`, `/`, `//`, `%` (left to right), then `+`, `-`. Use parentheses `()` to force a different order.

````python
print(2 + 3 * 4)        # multiplication happens first
print((2 + 3) * 4)      # parentheses force addition first
````
