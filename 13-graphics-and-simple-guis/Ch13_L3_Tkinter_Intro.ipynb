### 1. Your First Window

```python
import tkinter as tk

root = tk.Tk()
root.title("My First Window")

label = tk.Label(root, text="Hello, World!")
label.pack()

root.mainloop()
```

### 2. A Button That Does Something
This is **event-driven programming**: instead of code running top-to-bottom once, a function (the "callback") runs **whenever** the button is clicked — possibly many times.

```python
import tkinter as tk

count = 0

def on_click():
    global count
    count += 1
    label.config(text=f"Clicked {count} times")

root = tk.Tk()
label = tk.Label(root, text="Clicked 0 times")
label.pack()

button = tk.Button(root, text="Click me", command=on_click)
button.pack()

root.mainloop()
```
