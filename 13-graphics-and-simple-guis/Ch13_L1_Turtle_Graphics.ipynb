### 1. The Basics
`turtle` controls a little "pen" that moves around a window, drawing a line as it goes.

```python
import turtle

t = turtle.Turtle()
t.forward(100)
t.right(90)
t.forward(100)

turtle.done()   # keeps the window open until you close it
```

### 2. Turning and Color

```python
import turtle

t = turtle.Turtle()
t.pencolor("blue")
for _ in range(4):
    t.forward(100)
    t.left(90)     # a square, using the for loop from Chapter 8!

turtle.done()
```

### 3. Any Regular Polygon With a Loop
The turning angle for a regular polygon with `n` sides is always `360 / n`.

```python
import turtle

t = turtle.Turtle()
n = 6   # a hexagon
angle = 360 / n
for _ in range(n):
    t.forward(80)
    t.left(angle)

turtle.done()
```
