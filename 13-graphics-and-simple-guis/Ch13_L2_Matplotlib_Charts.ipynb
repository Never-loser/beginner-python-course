# Simple Charts with Matplotlib

### 1. A Basic Bar Chart

```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
hours_slept = [7, 6.5, 8, 5, 7.5]

plt.bar(days, hours_slept)
plt.title("Hours Slept This Week")
plt.xlabel("Day")
plt.ylabel("Hours")
plt.show()
```

### 2. A Basic Line Chart

```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
temperature = [22, 24, 21, 19, 23]

plt.plot(days, temperature, marker="o")
plt.title("Temperature This Week")
plt.xlabel("Day")
plt.ylabel("°C")
plt.show()
```
