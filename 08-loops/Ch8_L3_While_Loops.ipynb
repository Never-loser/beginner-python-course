# `while` Loops

### 1. Basic `while` Loop
A `while` loop repeats **as long as** its condition stays `True`. Unlike `for`, you control the repetition yourself — which means you must remember to update something inside the loop, or it never stops.

````python
count = 0
while count < 5:
    print(count)
    count += 1     # without this line, the loop never ends!
````

### 2. The Danger of Infinite Loops
If the condition never becomes `False`, the loop runs forever and the kernel gets stuck — you'll need to **Interrupt the Kernel** (from Chapter 1 Lesson 3) to stop it.
