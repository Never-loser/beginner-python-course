# Looping Over Lists, Strings, and Dictionaries

### 1. Looping Over a List Directly

```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

### 2. `enumerate()` — Getting the Index Too

```python
fruits = ["apple", "banana", "cherry"]
for index, fruit in enumerate(fruits):
    print(index, "->", fruit)
```

### 3. Looping Over a String

```python
word = "hello"
for letter in word:
    print(letter)
```

### 4. Looping Over a Dictionary (Recap From Chapter 7)

```python
student = {"name": "Sara", "age": 21}
for key, value in student.items():
    print(key, "->", value)
```
