# `break`, `continue`, and Nested Loops

### 1. `break` — Exit the Loop Immediately

```python
numbers = [3, 7, 12, 5, 9]
for n in numbers:
    if n > 10:
        print("Found a number over 10:", n)
        break
    print("checked:", n)
```

### 2. `continue` — Skip to the Next Iteration

```python
numbers = [1, 2, 3, 4, 5, 6]
for n in numbers:
    if n % 2 == 0:
        continue    # skip even numbers
    print(n)
```

### 3. Nested Loops
A loop inside another loop. The inner loop completes **fully** for each single pass of the outer loop. Important: `break` only exits the **innermost** loop it's written in — not both.

```python
for outer in range(3):
    for inner in range(2):
        print(f"outer={outer}, inner={inner}")
```
