# `for` Loops and `range()`

### 1. Basic `for` Loop
A `for` loop repeats a block of code once for each item in a sequence.

````python
for i in range(5):
    print(i)
````

### 2. `range()` Variations
| Call | Produces |
|---|---|
| `range(5)` | `0, 1, 2, 3, 4` |
| `range(2, 6)` | `2, 3, 4, 5` |
| `range(10, 0, -2)` | `10, 8, 6, 4, 2` |

Note: the **stop** value is always **excluded**.

````python
for i in range(2, 6):
    print(i)
print("---")
for i in range(10, 0, -2):
    print(i)
````
