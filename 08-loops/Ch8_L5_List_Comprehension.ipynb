# List Comprehension

### 1. The Pattern We're Replacing
A very common pattern: build an empty list, loop, and `.append()` to it.

```python
squares = []
for n in range(1, 6):
    squares.append(n ** 2)
print(squares)
```

### 2. The Same Thing, as a List Comprehension
`[expression for item in iterable]` — does the exact same job in one line.

```python
squares = [n ** 2 for n in range(1, 6)]
print(squares)
```

### 3. Adding a Filter With `if`

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
evens = [n for n in numbers if n % 2 == 0]
print(evens)
```
