# Lists — Definition and Basics

### 1. What is a List?
A **list** is an ordered collection of values, written with square brackets `[ ]`, separated by commas. Unlike a single string or number, a list can hold **many values of mixed types** at once.

````python
groceries = ["milk", "eggs", "bread"]
mixed = ["Sara", 21, 1.75, True]
print(groceries)
print(mixed)
print(len(groceries))
````

### 2. Accessing Elements
Lists use the exact same indexing rules as strings (Chapter 4): index `0` is first, `-1` is last.

````python
groceries = ["milk", "eggs", "bread"]
print(groceries[0])
print(groceries[-1])
````

### 3. Key Difference From Strings: Lists Are Mutable
Recall from Chapter 4 that strings **cannot** be changed in place. Lists **can** — you can assign directly to an index.

````python
groceries = ["milk", "eggs", "bread"]
groceries[1] = "cheese"    # this works! (would crash for a string)
print(groceries)
````
