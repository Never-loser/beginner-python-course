# Tuples and How They Differ From Lists

### 1. What is a Tuple?
A **tuple** looks like a list but uses parentheses `( )` instead of square brackets, and — most importantly — it is **immutable**: once created, it cannot be changed.

```python
point = (35.7, 51.4)   # example: latitude, longitude
print(point)
print(type(point))
print(point[0])        # indexing works just like lists
```

### 2. Tuples Cannot Be Changed

```python
# This will crash on purpose — read the error message
point = (35.7, 51.4)
point[0] = 40.0
```

### 3. Tuple Packing and Unpacking

```python
# Packing: putting values into a tuple
person = ("Sara", 21, "Tehran")

# Unpacking: pulling them back out into separate variables
name, age, city = person
print(name, age, city)
```

### 4. When to Use a Tuple Instead of a List
Use a tuple when the collection represents a **fixed** thing that shouldn't change — like a coordinate, a date, or a fixed-size record — and a list when you expect to add, remove, or reorder items.
