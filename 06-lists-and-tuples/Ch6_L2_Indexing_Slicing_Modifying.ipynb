### 1. Slicing Works Just Like Strings

````python
numbers = [10, 20, 30, 40, 50]
print(numbers[1:3])
print(numbers[::-1])    # reversed copy
````

### 2. Adding and Removing Elements
| Method | What it does |
|---|---|
| `.append(x)` | adds `x` to the end |
| `.insert(i, x)` | inserts `x` at index `i` |
| `.remove(x)` | removes the **first** matching value `x` (error if not found) |
| `.pop(i)` | removes **and returns** the item at index `i` (last item if `i` omitted) |
| `del list[i]` | removes the item at index `i` (no return value) |

````python
cart = ["apple", "banana"]
cart.append("cherry")
print(cart)
cart.insert(1, "kiwi")
print(cart)
cart.remove("banana")
print(cart)
removed = cart.pop(0)
print(removed, cart)
````
