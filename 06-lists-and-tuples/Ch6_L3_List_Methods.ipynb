# List Methods

### 1. Sorting

````python
numbers = [5, 2, 9, 1]
numbers.sort()             # sorts IN PLACE, changes the original list
print(numbers)
numbers.sort(reverse=True)
print(numbers)
````

### 2. `sorted()` — the Non-Destructive Version
Unlike `.sort()`, the built-in function `sorted()` returns a **new** sorted list and leaves the original unchanged.

````python
numbers = [5, 2, 9, 1]
new_list = sorted(numbers)
print(numbers)     # unchanged
print(new_list)    # sorted copy
````

### 3. Copying a List Correctly
Recall from Lesson 1: `list_b = list_a` does **not** make a copy — both names point to the same list. Use `.copy()` (or `list(...)`) to make a real, independent copy.

````python
original = [1, 2, 3]
real_copy = original.copy()
real_copy.append(4)
print(original)     # unaffected this time
print(real_copy)
````

### 4. Sorting With a `key`
The `key` parameter lets you sort by something other than the default order — for example, by length.

````python
words = ["python", "hi", "programming", "ok"]
words.sort(key=len)
print(words)
````
